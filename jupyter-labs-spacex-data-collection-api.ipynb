Below we will define a series of helper functions that will help us use the API to extract information using identification numbers in the launch data.

From the <code>rocket</code> column we would like to learn the booster name.


```python
API_TIMEOUT = (1.5, 2.5)
FALLBACK_ROCKET_NAMES = {
    '5e9d0d95eda69955f709d1eb': 'Falcon 1',
    '5e9d0d95eda69973a809d1ec': 'Falcon 9',
    '5e9d0d95eda69974db09d1ed': 'Falcon Heavy',
}

def _fetch_api_records(resource, identifiers):
    unique_ids = list(dict.fromkeys(
        identifier for identifier in identifiers
        if isinstance(identifier, str) and identifier
    ))
    if not unique_ids:
        return {}

    url = f'https://api.spacexdata.com/v4/{resource}'
    try:
        response = requests.get(url, timeout=API_TIMEOUT)
        response.raise_for_status()
        records = response.json()
        if not isinstance(records, list):
            raise ValueError(f'Expected a list from {url}')
    except (requests.exceptions.RequestException, ValueError) as exc:
        print(
            f'Warning: {resource} API lookup failed ({exc}); '
            'using the course dataset fallback where available.'
        )
        return {}

    records_by_id = {}
    for record in records:
        record_id = record.get('id') or record.get('_id')
        if record_id:
            records_by_id[record_id] = record
    missing_ids = [identifier for identifier in unique_ids if identifier not in records_by_id]
    if missing_ids:
        print(
            f'Warning: {len(missing_ids)} {resource} record(s) were absent from the API response; '
            'using the course dataset fallback where available.'
        )
    return {identifier: records_by_id[identifier] for identifier in unique_ids if identifier in records_by_id}

def _fallback_value(launch_date, column):
    launch_record = ENRICHMENT_BY_DATE.get(launch_date, {})
    return launch_record.get(column)

def _prefer_api_value(api_record, api_column, launch_date, fallback_column):
    value = api_record.get(api_column) if api_record else None
    if value is None or pd.isna(value):
        return _fallback_value(launch_date, fallback_column)
    return value

def getBoosterVersion(data):
    rocket_ids = data['rocket'].tolist()
    lookup = _fetch_api_records('rockets', rocket_ids)
    for rocket_id, launch_date in zip(rocket_ids, data['date']):
        name = _prefer_api_value(
            lookup.get(rocket_id), 'name', launch_date, 'BoosterVersion'
        )
        if name is None:
            name = FALLBACK_ROCKET_NAMES.get(rocket_id)
        BoosterVersion.append(name)
```

From the <code>launchpad</code> we would like to know the name of the launch site being used, the logitude, and the latitude.


```python
def getLaunchSite(data):
    launchpad_ids = data['launchpad'].tolist()
    lookup = _fetch_api_records('launchpads', launchpad_ids)
    for launchpad_id, launch_date in zip(launchpad_ids, data['date']):
        launchpad_data = lookup.get(launchpad_id)
        Longitude.append(
            _prefer_api_value(launchpad_data, 'longitude', launch_date, 'Longitude')
        )
        Latitude.append(
            _prefer_api_value(launchpad_data, 'latitude', launch_date, 'Latitude')
        )
        LaunchSite.append(
            _prefer_api_value(launchpad_data, 'name', launch_date, 'LaunchSite')
        )
```

From the <code>payload</code> we would like to learn the mass of the payload and the orbit that it is going to.


```python
def getPayloadData(data):
    payload_ids = data['payloads'].tolist()
    lookup = _fetch_api_records('payloads', payload_ids)
    for payload_id, launch_date in zip(payload_ids, data['date']):
        payload_data = lookup.get(payload_id)
        PayloadMass.append(
            _prefer_api_value(payload_data, 'mass_kg', launch_date, 'PayloadMass')
        )
        Orbit.append(
            _prefer_api_value(payload_data, 'orbit', launch_date, 'Orbit')
        )
```

From <code>cores</code> we would like to learn the outcome of the landing, the type of the landing, number of flights with that core, whether gridfins were used, wheter the core is reused, wheter legs were used, the landing pad used, the block of the core which is a number used to seperate version of cores, the number of times this specific core has been reused, and the serial of the core.


```python
def getCoreData(data):
    cores = data['cores'].tolist()
    core_ids = [core.get('core') for core in cores]
    lookup = _fetch_api_records('cores', core_ids)
    for core, core_id, launch_date in zip(cores, core_ids, data['date']):
        core_data = lookup.get(core_id)
        Block.append(
            _prefer_api_value(core_data, 'block', launch_date, 'Block')
        )
        ReusedCount.append(
            _prefer_api_value(core_data, 'reuse_count', launch_date, 'ReusedCount')
        )
        Serial.append(
            _prefer_api_value(core_data, 'serial', launch_date, 'Serial')
        )
        Outcome.append(f"{core.get('landing_success')} {core.get('landing_type')}")
        Flights.append(core.get('flight'))
        GridFins.append(core.get('gridfins'))
        Reused.append(core.get('reused'))
        Legs.append(core.get('legs'))
        LandingPad.append(core.get('landpad'))
```
